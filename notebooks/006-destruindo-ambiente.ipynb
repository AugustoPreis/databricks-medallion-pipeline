# 006 - Destruindo o Ambiente

Este notebook remove todos os schemas e volumes criados pelo notebook `001`,
incluindo todas as tabelas Delta e arquivos CSV armazenados.

Use-o para resetar o ambiente e reexecutar o pipeline do zero.

> **Atenção:** esta operação é irreversível. Todos os dados serão apagados.

## 1. Remover todos os schemas

`CASCADE` garante que tabelas, views e volumes internos sejam removidos junto com o schema.

```python
%%sql
DROP SCHEMA IF EXISTS workspace.bronze  CASCADE;
DROP SCHEMA IF EXISTS workspace.silver  CASCADE;
DROP SCHEMA IF EXISTS workspace.gold    CASCADE;
DROP SCHEMA IF EXISTS workspace.landing CASCADE;
```

## 2. Confirmar que os schemas foram removidos

```python
%%sql
SHOW SCHEMAS IN workspace;
```
